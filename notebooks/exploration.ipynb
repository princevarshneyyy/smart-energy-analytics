This notebook provides quick offline data inspection and visualization of smart meter time-series patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load cleaned featured data
csv_path = Path('../data/processed/featured_energy_data.csv')
if csv_path.exists():
    df = pd.read_csv(csv_path)
    print(f"Loaded {len(df):,} records successfully!")
    print(df.head())
else:
    print("Featured data not found. Please run run_pipeline.py first.")
```

```python
# Diurnal hourly consumption curves per consumer sector
if 'df' in locals():
    plt.figure(figsize=(10, 5))
    sns.lineplot(data=df, x='hour', y='energy_kwh', hue='consumer_type', marker='o')
    plt.title("Diurnal (24-Hour) Load Profiles Across Consumer Sectors")
    plt.xlabel("Hour of Day (0 to 23)")
    plt.ylabel("Active Power Consumption (kWh)")
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```
